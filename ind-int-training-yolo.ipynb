```python
from ultralytics import YOLO
from pathlib import Path
```

В прошлом коде path сохраняет путь по-прошлому принципу, здесь же приходится менять путь из-за нового создания блокнота -> создаем новый data.yaml, -> "output" /kaggle/working/data.yaml

```python
import yaml
from pathlib import Path

old_yaml = "/kaggle/input/datasets/nikita041/my-dataset/dataset/data.yaml"
with open(old_yaml, encoding="utf-8") as f:
    cfg = yaml.safe_load(f)

cfg["path"] = "/kaggle/input/datasets/nikita041/my-dataset/dataset"

new_yaml = "/kaggle/working/data.yaml"
with open(new_yaml, "w", encoding="utf-8") as f:
    yaml.safe_dump(cfg, f, allow_unicode=True)

print(open(new_yaml, encoding="utf-8").read())
```

```python
model = YOLO("yolo11n.pt")

results = model.train(
    data="/kaggle/working/data.yaml",   # наша "записка" с путями и буквами
    epochs=100,        # сколько раз прокрутить весь датасет
    imgsz=640,         # к какому размеру приводить картинки (640x640 — стандарт YOLO)
    batch=32,          # сколько картинок обрабатывать за один заход
    device=0,          # 0 = использовать GPU (не CPU!)
    patience=20,       # если 20 эпох подряд нет улучшения — остановиться раньше
    project="runs",    # куда складывать результаты
    name="dactyl",     # имя папки с этим запуском
)
```
